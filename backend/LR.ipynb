```python
def hypothesis_function(x,weight=0,bias=0):
    y_pred = weight * x + bias
    #   0 = 0 * 10 + 0
    return y_pred
```

```python
def cost_function(x,y):
    cost = (1/2*rows)*sum((hypothesis_function(x,y)-y)**2)
    return cost
```

```python
def gradient_decent(alpha,x,y,weight, bias):
    grad_weight = (1/rows)*sum((hypothesis_function(x,y)-y)*x)
    weight = weight - alpha * grad_weight
    
    grad_bias = (1/rows)* sum(hypothesis_function(x,y)-y)
    bias = bias - alpha*grad_bias
    
    return (weight,bias)
```

```python
def linear_regression(x,y):
    weight = .18
    bias = 0
    alpha = 0.001
    iter = 10
    for i in range(iter):
        y_pred = hypothesis_function(x,weight,bias)
        cost = cost_function(x,y)
        weight, bias = gradient_decent(alpha,x,y,weight,bias)
        
    return(weight,bias)
```
